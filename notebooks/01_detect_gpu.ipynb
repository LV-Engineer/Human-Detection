```python
from ultralytics import YOLO
from PIL import Image

def use_model(model_name: str, filepath: str) -> None:
    model = YOLO(model_name)
    result = model(filepath, classes=[0], conf=0.4)[0]

    for box in result.boxes:
        print(f"person conf={box.conf.item():.2f} xyxy={box.xyxy[0].tolist()}")
        
    display(Image.fromarray(result.plot()[..., ::-1]))
```

```python
import time
import cv2

frame = cv2.imread("/workspace/data/photo/47.JPG")
print(frame.shape)

for _ in range(3):
    model(frame, device=0, classes=[0], verbose=False)

runs = 20
start = time.perf_counter()
for _ in range(runs):
    model(frame, device=0, classes=[0], verbose=False)
ms = (time.perf_counter() - start) / runs * 1000

print(f"{ms:.1f} ms/кадр, {1000 / ms:.1f} FPS")

```
